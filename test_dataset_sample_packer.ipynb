```python
# run_locally_on_your_pc.py
import numpy as np
import medmnist
from medmnist import PathMNIST
from torchvision.transforms import v2
import torch

tf = v2.Compose([
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
])

test_dataset = PathMNIST(root="./data/",split="val",transform=tf,download=True,size=64)

num_samples = 256
random_indices = np.random.choice(len(test_dataset), size=num_samples, replace=False)

sample_images = []
sample_labels = []

for idx in random_indices:
    img_array = test_dataset.imgs[idx]
    label_array = test_dataset.labels[idx]
    
    sample_images.append(img_array)
    sample_labels.append(label_array)

np.savez_compressed(
    "val_samples.npz", 
    images=np.array(sample_images), 
    labels=np.array(sample_labels)
)

print(f"Created 'val_samples.npz' containing {num_samples} samples successfully!")
```
